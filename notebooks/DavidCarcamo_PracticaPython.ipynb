# Ejercicio 1: Clasificación de edad

**Enunciado:** determinar si una persona es niño, adolescente, adulto o adulto mayor según su edad.

**Lógica:** se define una edad directamente en el código y se evalúa usando condiciones `if`, `elif` y `else`. El primer rango que se cumple determina la categoría correspondiente.

```python
def clasificar_edad(edad):
    if edad < 13:
        return "Niño"
    elif edad <= 17:
        return "Adolescente"
    elif edad <= 59:
        return "Adulto"
    else:
        return "Adulto mayor"


edad = 21
categoria = clasificar_edad(edad)

print(f"Edad: {edad} años")
print(f"Categoría: {categoria}")
```

# Ejercicio 2: Conversor de temperatura

**Enunciado:** convertir una temperatura de grados Celsius a Fahrenheit utilizando la fórmula correspondiente.

**Lógica:** se define una temperatura en grados Celsius directamente en el código. Luego se aplica la fórmula de conversión `F = (C × 9/5) + 32` para obtener la temperatura equivalente en grados Fahrenheit y se muestra el resultado con un mensaje claro.

```python
def convertir_fahrenheit(celsius):
    fahrenheit = (celsius * 9/5) + 32
    return fahrenheit


temperatura_celsius = 25
temperatura_fahrenheit = convertir_fahrenheit(temperatura_celsius)

print(f"Temperatura en Celsius: {temperatura_celsius} °C")
print(f"Temperatura en Fahrenheit: {temperatura_fahrenheit} °F")
```

# Ejercicio 3: Calculadora básica

**Enunciado:** realizar una operación matemática entre dos números. Las operaciones permitidas son suma, resta, multiplicación y división. Además, se debe validar que no se intente dividir entre cero.

**Lógica:** se definen dos números y una operación matemática directamente en el código. Mediante una estructura `if`, `elif` y `else` se identifica la operación seleccionada y se ejecuta el cálculo correspondiente. En el caso de la división, se verifica previamente que el segundo número no sea cero para evitar errores.

```python
def calculadora(numero1, numero2, operacion):
    if operacion == "suma":
        return numero1 + numero2

    elif operacion == "resta":
        return numero1 - numero2

    elif operacion == "multiplicacion":
        return numero1 * numero2

    elif operacion == "division":
        if numero2 == 0:
            return "Error: no se puede dividir entre cero"
        return numero1 / numero2

    else:
        return "Operación no válida"


numero1 = 20
numero2 = 5
operacion = "division"

resultado = calculadora(numero1, numero2, operacion)

print(f"Número 1: {numero1}")
print(f"Número 2: {numero2}")
print(f"Operación: {operacion}")
print(f"Resultado: {resultado}")
```

# Ejercicio 8: Lista de números positivos y negativos

**Enunciado:** contar cuántos números positivos, negativos y ceros existen dentro de una lista de diez números.

**Lógica:** se define una lista con diez números enteros. Mediante un ciclo `for`, se recorre cada elemento de la lista y se evalúa si es mayor que cero, menor que cero o igual a cero. Dependiendo del caso, se incrementa el contador correspondiente. Al finalizar, se muestra un resumen con la cantidad de números positivos, negativos y ceros encontrados.

```python
numeros = [15, -8, 0, 23, -12, 7, 0, -3, 18, -5]

positivos = 0
negativos = 0
ceros = 0

for numero in numeros:
    if numero > 0:
        positivos += 1
    elif numero < 0:
        negativos += 1
    else:
        ceros += 1

print(f"Lista de números: {numeros}")
print(f"Cantidad de positivos: {positivos}")
print(f"Cantidad de negativos: {negativos}")
print(f"Cantidad de ceros: {ceros}")
```

# Ejercicio 11: Función para calcular descuento

**Enunciado:** crear una función llamada `calcular_descuento` que reciba el precio original de un producto y el porcentaje de descuento. La función debe calcular y retornar el precio final después de aplicar el descuento.

**Lógica:** se crea una función que recibe dos parámetros: el precio original y el porcentaje de descuento. Primero se calcula el valor del descuento y luego se resta al precio original para obtener el precio final. Finalmente, la función retorna el resultado para ser utilizado fuera de ella.

**Datos de entrada:** precio original y porcentaje de descuento.

**Dato de salida:** precio final después de aplicar el descuento.

```python
def calcular_descuento(precio_original, porcentaje_descuento):
    descuento = precio_original * (porcentaje_descuento / 100)
    precio_final = precio_original - descuento
    return precio_final


precio = 1500
descuento = 20

precio_final = calcular_descuento(precio, descuento)

print(f"Precio original: L. {precio}")
print(f"Descuento aplicado: {descuento}%")
print(f"Precio final: L. {precio_final}")
```

# Ejercicio 12: Validación de contraseña

**Enunciado:** validar si una contraseña cumple con los requisitos mínimos de seguridad. Debe tener al menos 8 caracteres, contener al menos un número, una letra mayúscula y una letra minúscula.

```python
contrasena = "BigData2025"

tiene_longitud = len(contrasena) >= 8
tiene_numero = any(caracter.isdigit() for caracter in contrasena)
tiene_mayuscula = any(caracter.isupper() for caracter in contrasena)
tiene_minuscula = any(caracter.islower() for caracter in contrasena)

if tiene_longitud and tiene_numero and tiene_mayuscula and tiene_minuscula:
    resultado = "Contraseña válida"
else:
    resultado = "Contraseña no válida"

print(f"Contraseña evaluada: {contrasena}")
print(resultado)
```

# Ejercicio 13: Sistema simple de inventario

**Enunciado:** crear un sistema simple de inventario que almacene productos con código, nombre, precio y cantidad disponible. Mostrar todos los productos, buscar un producto por código y calcular el valor total del inventario.

**Lógica:** se utiliza una lista de diccionarios para almacenar la información de cada producto. Primero se muestran todos los productos registrados. Luego se busca un producto utilizando su código. Finalmente, se calcula el valor total del inventario multiplicando el precio por la cantidad disponible de cada producto y sumando todos los resultados.

```python
productos = [
    {"codigo": "P001", "nombre": "Laptop", "precio": 18000, "cantidad": 5},
    {"codigo": "P002", "nombre": "Mouse", "precio": 350, "cantidad": 20},
    {"codigo": "P003", "nombre": "Teclado", "precio": 800, "cantidad": 10},
    {"codigo": "P004", "nombre": "Monitor", "precio": 4500, "cantidad": 8}
]

print("LISTA DE PRODUCTOS")
print("-" * 60)

for producto in productos:
    print(
        f"Código: {producto['codigo']} | "
        f"Nombre: {producto['nombre']} | "
        f"Precio: L. {producto['precio']} | "
        f"Cantidad: {producto['cantidad']}"
    )

codigo_buscar = "P003"

print("\nBÚSQUEDA DE PRODUCTO")
print("-" * 60)

for producto in productos:
    if producto["codigo"] == codigo_buscar:
        print(
            f"Producto encontrado: {producto['nombre']} "
            f"(Precio: L. {producto['precio']}, "
            f"Cantidad: {producto['cantidad']})"
        )

valor_total = 0

for producto in productos:
    valor_total += producto["precio"] * producto["cantidad"]

print("\nVALOR TOTAL DEL INVENTARIO")
print("-" * 60)
print(f"L. {valor_total}")
```

# Ejercicio 14: Registro de ventas

**Enunciado:** crear un registro de ventas donde cada venta tenga cliente, producto, cantidad y precio unitario. Calcular el total por cada venta, el total general vendido e identificar el cliente con la compra más alta.

**Lógica:** se utiliza una lista de diccionarios para almacenar las ventas. Cada diccionario representa una venta con su cliente, producto, cantidad y precio unitario. Luego se recorre la lista para calcular el total de cada venta multiplicando la cantidad por el precio unitario. También se acumula el total general vendido y se identifica la venta con el monto más alto.

```python
ventas = [
    {"cliente": "Carlos Méndez", "producto": "Laptop", "cantidad": 1, "precio_unitario": 18000},
    {"cliente": "María López", "producto": "Mouse", "cantidad": 3, "precio_unitario": 350},
    {"cliente": "Ana Rodríguez", "producto": "Monitor", "cantidad": 2, "precio_unitario": 4500},
    {"cliente": "José Martínez", "producto": "Teclado", "cantidad": 4, "precio_unitario": 800}
]

total_general = 0
compra_mas_alta = ventas[0]

print("REGISTRO DE VENTAS")
print("-" * 70)

for venta in ventas:
    total_venta = venta["cantidad"] * venta["precio_unitario"]
    venta["total"] = total_venta
    total_general += total_venta

    if total_venta > compra_mas_alta["cantidad"] * compra_mas_alta["precio_unitario"]:
        compra_mas_alta = venta

    print(
        f"Cliente: {venta['cliente']} | "
        f"Producto: {venta['producto']} | "
        f"Cantidad: {venta['cantidad']} | "
        f"Precio unitario: L. {venta['precio_unitario']} | "
        f"Total: L. {total_venta}"
    )

print("\nTOTAL GENERAL VENDIDO")
print("-" * 70)
print(f"L. {total_general}")

print("\nCLIENTE CON LA COMPRA MÁS ALTA")
print("-" * 70)
print(f"Cliente: {compra_mas_alta['cliente']}")
print(f"Producto: {compra_mas_alta['producto']}")
print(f"Total de compra: L. {compra_mas_alta['total']}")
```

# Ejercicio 15: Limpieza básica de texto

**Enunciado:** realizar una limpieza básica de una frase eliminando espacios innecesarios, convirtiendo el texto a minúsculas, reemplazando espacios dobles por espacios simples y contando la cantidad de palabras contenidas en la frase.

**Lógica:** se define una frase con espacios adicionales al inicio, al final y entre algunas palabras. Posteriormente se aplican diferentes métodos de cadena para limpiar el texto. Primero se eliminan los espacios externos, luego se convierte el texto a minúsculas y se sustituyen los espacios dobles por espacios simples. Finalmente, se divide la frase en palabras para contabilizar cuántas contiene.

**Métodos utilizados:**

- `strip()`: elimina espacios en blanco al inicio y al final de una cadena.
- `lower()`: convierte todos los caracteres del texto a minúsculas.
- `replace()`: reemplaza una cadena por otra dentro del texto.
- `split()`: divide una cadena en una lista de palabras utilizando los espacios como separadores.

```python
frase = "   Python   es  muy UTILIZADO en   Big Data   "

print("Texto original:")
print(f"'{frase}'")

texto_limpio = frase.strip()
texto_limpio = texto_limpio.lower()

while "  " in texto_limpio:
    texto_limpio = texto_limpio.replace("  ", " ")

palabras = texto_limpio.split()
cantidad_palabras = len(palabras)

print("\nTexto limpio:")
print(f"'{texto_limpio}'")

print(f"\nCantidad de palabras: {cantidad_palabras}")
print(f"Lista de palabras: {palabras}")
```

**Lógica:** se crea un DataFrame utilizando pandas con información de ventas. Se calcula el valor total de cada venta multiplicando la cantidad por el precio unitario. Después se utilizan funciones de agrupación y agregación para obtener estadísticas de ventas por vendedor, producto y categoría. Finalmente, se ordenan los registros según el monto de venta.

# Ejercicio 18: Limpieza de datos con pandas

**Enunciado:** crear un DataFrame con información de clientes que contenga algunos datos incompletos. Posteriormente se deben identificar los valores nulos, contar cuántos existen por columna, reemplazar las ciudades vacías por "No especificada", eliminar los registros que no tengan correo electrónico y mostrar el DataFrame limpio.

**Lógica:** se crea un DataFrame con algunos valores faltantes utilizando `None`. Luego se emplean métodos de pandas para detectar y contabilizar los valores nulos. Después se reemplazan las ciudades vacías por un valor predeterminado y se eliminan los registros que no poseen correo electrónico. Finalmente se muestra el DataFrame resultante después de la limpieza.

**Enunciado:** crear un DataFrame con solicitudes de préstamo que incluya cliente, monto solicitado, plazo, ingreso mensual y estado de solicitud. Posteriormente se debe calcular el monto promedio solicitado, contar solicitudes por estado, mostrar únicamente los préstamos aprobados, identificar clientes con montos superiores a 50,000 y agrupar los datos por estado para calcular el promedio del monto solicitado.

**Lógica:** se crea un DataFrame utilizando pandas para almacenar información de solicitudes de préstamo. Luego se aplican funciones de análisis para obtener estadísticas generales, filtrar registros específicos y agrupar información según el estado de la solicitud.

```python
import pandas as pd

prestamos = {
    "Cliente": [
        "Carlos Méndez",
        "Ana López",
        "José Rivera",
        "María Torres",
        "Luis Martínez",
        "Sofía Hernández",
        "Pedro Castillo",
        "Gabriela Flores"
    ],
    "MontoSolicitado": [
        25000,
        60000,
        45000,
        80000,
        30000,
        55000,
        70000,
        40000
    ],
    "PlazoMeses": [
        24,
        48,
        36,
        60,
        24,
        48,
        60,
        36
    ],
    "IngresoMensual": [
        18000,
        35000,
        22000,
        45000,
        20000,
        38000,
        50000,
        25000
    ],
    "EstadoSolicitud": [
        "Aprobado",
        "Aprobado",
        "Pendiente",
        "Aprobado",
        "Rechazado",
        "Pendiente",
        "Aprobado",
        "Rechazado"
    ]
}

df_prestamos = pd.DataFrame(prestamos)

print("DataFrame de préstamos:")
display(df_prestamos)

monto_promedio = df_prestamos["MontoSolicitado"].mean()
print(f"\nMonto promedio solicitado: L. {monto_promedio:,.2f}")

print("\nCantidad de solicitudes por estado:")
display(df_prestamos["EstadoSolicitud"].value_counts())

print("\nPréstamos aprobados:")
display(df_prestamos[df_prestamos["EstadoSolicitud"] == "Aprobado"])

print("\nClientes con monto solicitado mayor a L. 50,000:")
display(df_prestamos[df_prestamos["MontoSolicitado"] > 50000])

print("\nPromedio del monto solicitado por estado:")
display(
    df_prestamos.groupby("EstadoSolicitud")["MontoSolicitado"].mean()
)
```

# Ejercicio 20: Mini análisis de datos tipo empresa

**Enunciado:** crear un DataFrame con información de empleados que incluya nombre, departamento, cargo, salario, años de experiencia y ciudad. Posteriormente se deben mostrar los primeros registros, obtener estadísticas salariales, calcular el salario promedio por departamento, identificar al empleado con el salario más alto, mostrar empleados con más de 5 años de experiencia, ordenar los empleados por salario y crear una nueva columna llamada NivelExperiencia.

**Lógica:** se crea un DataFrame utilizando pandas para almacenar información de empleados. Después se aplican diferentes métodos de análisis para obtener estadísticas descriptivas, agrupar información por departamento, filtrar registros específicos y ordenar los datos. Finalmente, se crea una nueva columna que clasifica a cada empleado según sus años de experiencia.

```python
import pandas as pd

empleados = {
    "Nombre": [
        "David Carcamo",
        "María López",
        "Carlos Méndez",
        "Ana Rodríguez",
        "José Rivera",
        "Sofía Hernández",
        "Luis Martínez",
        "Gabriela Flores"
    ],
    "Departamento": [
        "Tecnología",
        "Ventas",
        "Finanzas",
        "Recursos Humanos",
        "Tecnología",
        "Ventas",
        "Finanzas",
        "Tecnología"
    ],
    "Cargo": [
        "Desarrollador",
        "Vendedora",
        "Analista Financiero",
        "Coordinadora RRHH",
        "Administrador de Sistemas",
        "Supervisora de Ventas",
        "Contador",
        "Ingeniera de Software"
    ],
    "Salario": [
        35000,
        28000,
        32000,
        26000,
        40000,
        38000,
        30000,
        45000
    ],
    "AniosExperiencia": [
        3,
        6,
        4,
        8,
        10,
        7,
        5,
        9
    ],
    "Ciudad": [
        "Tegucigalpa",
        "San Pedro Sula",
        "La Ceiba",
        "Comayagua",
        "Danlí",
        "Tegucigalpa",
        "Juticalpa",
        "San Pedro Sula"
    ]
}

df_empleados = pd.DataFrame(empleados)

print("Primeros registros del DataFrame:")
display(df_empleados.head())

print("\nEstadísticas generales de salarios:")
display(df_empleados["Salario"].describe())

print("\nSalario promedio por departamento:")
display(df_empleados.groupby("Departamento")["Salario"].mean())

print("\nEmpleado con salario más alto:")
display(df_empleados.loc[[df_empleados["Salario"].idxmax()]])

print("\nEmpleados con más de 5 años de experiencia:")
display(df_empleados[df_empleados["AniosExperiencia"] > 5])

print("\nEmpleados ordenados por salario (de mayor a menor):")
display(df_empleados.sort_values(by="Salario", ascending=False))

def clasificar_experiencia(anios):
    if anios <= 2:
        return "Junior"
    elif anios <= 5:
        return "Intermedio"
    else:
        return "Senior"

df_empleados["NivelExperiencia"] = df_empleados["AniosExperiencia"].apply(
    clasificar_experiencia
)

print("\nDataFrame con la columna NivelExperiencia:")
display(df_empleados)
```
