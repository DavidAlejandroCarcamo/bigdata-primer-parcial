## Caso seleccionado: Netflix

Actualmente, Netflix es una de las plataformas de streaming más utilizadas en el mundo y representa uno de los ejemplos más conocidos del uso de Big Data en la industria del entretenimiento. La empresa utiliza grandes cantidades de información para analizar el comportamiento de sus usuarios y ofrecer recomendaciones personalizadas. Gracias al análisis de datos, Netflix puede mejorar la experiencia de cada persona, optimizar sus servicios y tomar decisiones importantes sobre el contenido que produce.

## 1. ¿Qué datos genera?

Netflix genera enormes cantidades de datos cada segundo. Algunos ejemplos son:

- Películas y series vistas por los usuarios
- Tiempo de reproducción
- Pausas, retrocesos y avances
- Calificaciones y preferencias
- Horarios de conexión
- Ubicación geográfica aproximada
- Tipo de dispositivo utilizado
- Búsquedas realizadas en la plataforma

Estos datos se generan continuamente por millones de usuarios en todo el mundo, produciendo un volumen masivo de información diaria.

## 2. ¿Quién los usa?

Los datos son utilizados principalmente por:

- Netflix
- Equipos de análisis de datos
- Ingenieros de software
- Sistemas de inteligencia artificial
- Equipos de marketing

Netflix utiliza estos datos para analizar comportamientos, mejorar recomendaciones, detectar tendencias y optimizar el rendimiento de la plataforma.

## 3. ¿Qué problema resuelve?

El análisis de Big Data permite a Netflix:

- Recomendar contenido personalizado
- Mejorar la experiencia del usuario
- Reducir la cancelación de suscripciones
- Optimizar la calidad del streaming
- Identificar contenido popular
- Tomar decisiones sobre nuevas producciones

Gracias al análisis de datos, Netflix logra mantener a los usuarios más satisfechos y aumentar el tiempo de uso de la plataforma.


## 4. ¿Por qué no bastaría Excel?

Excel no sería suficiente debido a:

- El enorme volumen de datos generado diariamente
- La velocidad en que se producen los datos
- La variedad de formatos de información
- La necesidad de análisis en tiempo real

Netflix maneja millones de usuarios simultáneamente, por lo que necesita tecnologías avanzadas de Big Data capaces de procesar información distribuida en múltiples servidores.
